# 2 Plotting a Function
Goal: Write a **function** that will plot $f(x) = x^2$ and $g(x) = \sqrt{x}$ on the same plot and return the $f(x)$ and $g(x)$ values as separate lists. Your plot needs to have the following:
- An appropriate figure size (`figsize=(10,8)` is recommended)
- Labeled x and y axis
- A title
- Different colors and linestyles for each function
- A legend
- Your domain should be $x \in [0, 10]$. *Hint: Use `np.linspace` to set a range of values for your functions.*

Your function should return:
- A list of $f(x)$ values
- A list of $g(x)$ values.

```python

def plot_f_and_g():
    """
    Plots the functions f(x) = x^2 and g(x) = sqrt(x) on the same graph.
    """
    # Domain of function
    x = np.linspace(0, 5)

    # f and g
    f_values = x**2
    g_values = np.sqrt(x)

    plt.figure(figsize=(10, 8))
    plt.plot(x, f_values, label='f(x) = x^2', color='blue', linestyle='-')
    plt.plot(x, g_values, label='g(x) = sqrt(x)', color='green', linestyle='-')

    # Labels & grid
    plt.title('Plots of f(x) = x^2 and g(x) = sqrt(x)', fontsize=16)
    plt.xlabel('x', fontsize=14)
    plt.ylabel('y', fontsize=14)
    plt.grid(True)

    # Display the plot & legend
    plt.legend(fontsize=12)
    plt.show()

    # Return the computed values
    return f_values.tolist(), g_values.tolist()

# Call function, print values
f_values, g_values = plot_f_and_g()
print(f_values[:10])
print(g_values[:10])
```

```python
def plot_f_and_g_new():
    """
    Plots the functions f(x) = x^2 and g(x) = sqrt(x) on the same graph.
  
    """
    # Domain of functions (limits on the functions only, not the grid or the lines)
    x = np.linspace(0, 2)

    # f & g
    f_values = x**2
    g_values = np.sqrt(x)

    plt.figure(figsize=(10, 8))
    plt.plot(x, f_values, label='f(x) = x^2', color='blue', linestyle='-')
    plt.plot(x, g_values, label='g(x) = sqrt(x)', color='green', linestyle='-')
    plt.axvline(x=2, color='red', linestyle='-', linewidth=2, label='Horizontal Line (x = 2)')
    plt.axhline(y=1, color='purple', linestyle='-', linewidth=2, label='Vertical Line (y = 1)')

    # Graph labels
    plt.title('Plots of f(x) = x^2 and g(x) = sqrt(x)', fontsize=16)
    plt.xlabel('x', fontsize=14)
    plt.ylabel('y', fontsize=14)

    # Limits and grid
    plt.xlim(0, 3)
    plt.ylim(0, 9)
    plt.grid(True)

    # Display the plot & legend
    plt.legend(fontsize=12,loc='upper right', bbox_to_anchor=(1.4, 1))
    plt.show()

    # Return the computed values
    return f_values.tolist(), g_values.tolist()

# Call function, print values
f_values, g_values = plot_f_and_g()
h_values = f_values + g_values
print(f_values[:10])
print(g_values[:10]) # The 10 ensures it only prints 10
print(h_values[:10])
```

# 3 Introduction to Subplots
Goal: Write a **module** (name it whatever you want, as long as it is professional) that contains two functions that do the following: 

Function 1: 
- Write a **function** that will plot two subplots side-by-side (horizontal).
- The **left** subplot will be: $h(x) = \cos(x)$.
- The **right** subplot will be $k(x) = \sin(x)$.
- Include a title, axis labels and a domain of $x \in [0, 2 \pi]$ (this should NOT be hardcoded)

Function 2:
- Exactly the same as your first function, **EXCEPT** the subplots will be on top of each other (vertical). 

```python
def plot_horizontal_subplots(x):
    """
    Plots two subplots side-by-side (horizontal layout):
      - Left subplot: h(x) = cos(x)
      - Right subplot: k(x) = sin(x)
    """
    # Given domain & # of points plotted for smoother curve
    x = np.linspace(0, 2 * np.pi, 500)

    # Functions
    h_values = np.cos(x)
    k_values = np.sin(x)

    # Subplots, arranged in 1 row, 2 columns
    fig, axes = plt.subplots(1, 2, figsize=(12, 6))

    # Plot h(x) = cos(x) on the left
    axes[0].plot(x, h_values, color='blue', label='h(x) = cos(x)')
    axes[0].set_title('h(x) = cos(x)', fontsize=14)
    axes[0].set_xlabel('x', fontsize=12)
    axes[0].set_ylabel('h(x)', fontsize=12)
    axes[0].grid(True)
    axes[0].legend(fontsize=10)

    # Plot k(x) = sin(x) on the right
    axes[1].plot(x, k_values, color='green', label='k(x) = sin(x)')
    axes[1].set_title('k(x) = sin(x)', fontsize=14)
    axes[1].set_xlabel('x', fontsize=12)
    axes[1].set_ylabel('k(x)', fontsize=12)
    axes[1].grid(True)
    axes[1].legend(fontsize=10)

    # Adjust layout and show the plot
    plt.tight_layout()
    plt.show()

def plot_vertical_subplots(x):
    """
    Plots the same two subplots, but stacked vertically
    """
    # Same domain
    x = np.linspace(0, 2 * np.pi, 500)

    # Same functions
    h_values = np.cos(x)
    k_values = np.sin(x)

    # This time two rows & one column
    fig, axes = plt.subplots(2, 1, figsize=(8, 10))

    # Plot h(x) = cos(x) on the top
    axes[0].plot(x, h_values, color='blue', label='h(x) = cos(x)')
    axes[0].set_title('h(x) = cos(x)', fontsize=14)
    axes[0].set_xlabel('x', fontsize=12)
    axes[0].set_ylabel('h(x)', fontsize=12)
    axes[0].grid(True)
    axes[0].legend(fontsize=10)

    # Plot k(x) = sin(x) on the bottom
    axes[1].plot(x, k_values, color='green', label='k(x) = sin(x)')
    axes[1].set_title('k(x) = sin(x)', fontsize=14)
    axes[1].set_xlabel('x', fontsize=12)
    axes[1].set_ylabel('k(x)', fontsize=12)
    axes[1].grid(True)
    axes[1].legend(fontsize=10)

    # Adjust layout and show the plot
    plt.tight_layout()
    plt.show()
```

Call your **first** function in the cell below.

```python
plot_horizontal_subplots(x)
```

Call your **second** function in the cell below.

```python
plot_vertical_subplots(x)
```

```python
x_extended = np.linspace(0, 4 * np.pi, 500)
plot_horizontal_subplots(x_extended)
plot_vertical_subplots(x_extended)
```

```python
# Importing all modules including one I made that simulates die rolls
import matplotlib.pyplot as plt
import numpy as np
import dice_roller as roll_dice

num_sets = 1000 # Number of rolls
dice_numbers = [1, 2, 3, 4, 5, 10] # Number of dies rolled

def compute_total_scores(num_sets, num_dice):
    """
    A function that calculates the score (i.e. the sum of rolled values) obtained from N dice, where N is a positive integer
    """
    total_scores = []
    for _ in range(num_sets):
        dice_throws = roll_dice.roll_dice(num_dice)
        total_scores.append(np.sum(dice_throws))
    return np.array(total_scores)

plt.figure(figsize=(15, 10))

for i, N in enumerate(dice_numbers):
    total_scores = compute_total_scores(num_sets, N) # For loop iterates over all dice numbers

    # Graphing results in histogram form of scores vs probability density (i.e. how probable each score is to occur), three rows & two columns
    plt.subplot(2, 3, i + 1)
    plt.hist(total_scores, bins=np.arange(N, 6 * N + 1) - 0.5, density=True, edgecolor='black')
    plt.xlabel('Total Score')
    plt.ylabel('Probability Density')
    plt.title(f'Score Distribution for N={N}')
    plt.grid(True)

    # Calculation section, outputs mean & standard deviation for each value of N
    mean_score = np.mean(total_scores)
    std_dev_score = np.std(total_scores, ddof=1)  # Sample standard deviation
    print(f"N = {N}")
    print(f"Mean Total Score: {mean_score}")
    print(f"Standard Deviation: {std_dev_score}")

plt.tight_layout() # For cosmetic purposes
plt.show()
```
